```python
%sql
CREATE OR REPLACE TABLE silver.customers
USING DELTA
AS
SELECT
    cid AS customer_id,
    INITCAP(TRIM(cname)) AS customer_name,
    TRIM(cnumber) AS phone_number,
    INITCAP(TRIM(caddress)) AS address
FROM bronze.customer_raw;
```

```python
%sql
SELECT *
FROM silver.customers;
```

```python
%sql
CREATE OR REPLACE TABLE silver.tailors
USING DELTA
AS
SELECT
    tailor_id,
    INITCAP(TRIM(tailor_name)) AS tailor_name,
    TRIM(phone_no) AS phone_number
FROM bronze.tailor_raw;
```

```python
%sql
SELECT *
FROM silver.tailors;
```

```python
%sql
CREATE OR REPLACE TABLE silver.garments
USING DELTA
AS
SELECT
    fid AS garment_id,
    INITCAP(TRIM(type_name)) AS garment_type,
    TRIM(fabric_required) AS fabric_required
FROM bronze.garment_type_raw;
```

```python
%sql
SELECT *
FROM silver.garments;
```

```python
%sql
CREATE OR REPLACE TABLE silver.measurements
USING DELTA
AS
SELECT
    CAST(mid AS INT) AS measurement_id,
    CAST(waist AS DECIMAL(5,2)) AS waist,
    CAST(hip AS DECIMAL(5,2)) AS hip,
    CAST(armlength AS DECIMAL(5,2)) AS arm_length,
    CAST(armround AS DECIMAL(5,2)) AS arm_round,
    CAST(blength AS DECIMAL(5,2)) AS body_length,
    customerid AS customer_id
FROM bronze.measurements_raw;
```

```python
%sql
SELECT *
FROM silver.measurements;
```

```python
%sql
CREATE OR REPLACE TABLE silver.orders
USING DELTA
AS
SELECT
    oid AS order_id,

    TO_DATE(
        deadline,
        'd-M-yy'
    ) AS deadline_date,

    CAST(price AS DECIMAL(10,2)) AS order_amount,

    fid AS garment_id,

    UPPER(TRIM(status)) AS order_status,

    tailor_id

FROM bronze.orders_raw;
```

```python
%sql
SELECT *
FROM silver.orders;
```

```python
%sql
CREATE OR REPLACE TABLE silver.payments
USING DELTA
AS
SELECT
    payment_id,
    INITCAP(TRIM(mode_of_payment)) AS payment_method,
    CAST(price AS DECIMAL(10,2)) AS payment_amount
FROM bronze.payments_raw;
```

```python
%sql
SELECT *
FROM silver.payments;
```

```python
%sql
CREATE OR REPLACE TABLE silver.payments
USING DELTA
AS
SELECT
    payment_id,

    CASE
        WHEN UPPER(TRIM(mode_of_payment)) IN ('GPAY', 'G-PAY', 'GOOGLE PAY')
            THEN 'GPay'

        WHEN UPPER(TRIM(mode_of_payment)) IN ('COD', 'CASH ON DELIVERY')
            THEN 'COD'

        WHEN UPPER(TRIM(mode_of_payment)) IN ('NETBANKING', 'NET BANKING')
            THEN 'NetBanking'

        WHEN UPPER(TRIM(mode_of_payment)) = 'CARD'
            THEN 'Card'

        WHEN UPPER(TRIM(mode_of_payment)) = 'UPI'
            THEN 'UPI'

        ELSE 'Other'
    END AS payment_method,

    CAST(price AS DECIMAL(10,2)) AS payment_amount

FROM bronze.payments_raw;
```

```python
%sql
CREATE OR REPLACE TABLE silver.order_status_history
USING DELTA
AS
SELECT
    log_id,
    oid AS order_id,
    UPPER(TRIM(status)) AS status,
    changed_on,
    updated_by_tailor AS tailor_id
FROM bronze.order_status_log_raw;
```

```python
%sql
CREATE OR REPLACE TABLE gold.dim_date
USING DELTA
AS
SELECT DISTINCT
    deadline_date AS date,
    YEAR(deadline_date) AS year,
    QUARTER(deadline_date) AS quarter,
    MONTH(deadline_date) AS month,
    MONTHNAME(deadline_date) AS month_name,
    DAY(deadline_date) AS day,
    DAYOFWEEK(deadline_date) AS day_of_week
FROM silver.orders
WHERE deadline_date IS NOT NULL;
```

```python
%sql
CREATE OR REPLACE TABLE gold.fact_orders
USING DELTA
AS
SELECT
    o.order_id,
    cod.customer_id,
    cod.measurement_id,
    o.garment_id,
    o.tailor_id,
    o.deadline_date,
    o.order_amount,
    o.order_status
FROM silver.orders o
LEFT JOIN silver.customer_order_details cod
    ON o.order_id = cod.order_id;
```

```python
%sql
SELECT *
FROM gold.fact_orders
ORDER BY order_id;
```

```python
%sql
CREATE OR REPLACE TABLE gold.fact_orders
USING DELTA
AS
SELECT
    o.order_id,
    cod.customer_id,
    cod.measurement_id,
    o.garment_id,
    o.tailor_id,
    o.deadline_date,
    o.order_amount,
    o.order_status,

    CASE
        WHEN o.order_status = 'COMPLETED'
            THEN 1
        ELSE 0
    END AS completed_flag,

    CASE
        WHEN o.order_status = 'PENDING'
            THEN 1
        ELSE 0
    END AS pending_flag

FROM silver.orders o

LEFT JOIN silver.customer_order_details cod
    ON o.order_id = cod.order_id;
```

```python
%sql
CREATE OR REPLACE TABLE gold.fact_payments
USING DELTA
AS
SELECT
    p.payment_id,
    cop.customer_id,
    cop.order_id,
    p.payment_method,
    p.payment_amount
FROM silver.payments p

LEFT JOIN silver.customer_order_payment cop
    ON p.payment_id = cop.payment_id;
```

```python
%sql
SELECT *
FROM gold.fact_payments
ORDER BY order_id;
```

```python
%sql
SELECT
    COUNT(*) AS total_orders,
    COUNT(DISTINCT order_id) AS unique_orders,
    SUM(order_amount) AS total_revenue,
    SUM(completed_flag) AS completed_orders,
    SUM(pending_flag) AS pending_orders
FROM gold.fact_orders;
```

```python
%sql
SELECT
    COUNT(*) AS total_orders,
    SUM(order_amount) AS total_revenue,
    SUM(completed_flag) AS completed_orders,
    SUM(pending_flag) AS pending_orders,
    ROUND(AVG(order_amount), 2) AS average_order_value
FROM gold.fact_orders;
```

```python
%sql
SELECT
    g.garment_type,
    COUNT(f.order_id) AS total_orders,
    ROUND(SUM(f.order_amount), 2) AS total_revenue,
    ROUND(AVG(f.order_amount), 2) AS average_order_value
FROM gold.fact_orders f
JOIN gold.dim_garment g
    ON f.garment_id = g.garment_id
GROUP BY g.garment_type
ORDER BY total_revenue DESC;
```

```python
%sql
SELECT
    YEAR(deadline_date) AS year,
    MONTH(deadline_date) AS month,
    MONTHNAME(deadline_date) AS month_name,
    COUNT(*) AS total_orders,
    ROUND(SUM(order_amount), 2) AS total_revenue
FROM gold.fact_orders
GROUP BY
    YEAR(deadline_date),
    MONTH(deadline_date),
    MONTHNAME(deadline_date)
ORDER BY
    year,
    month;
```

```python
%sql
SELECT
    payment_method,
    COUNT(*) AS total_payments,
    ROUND(SUM(payment_amount), 2) AS total_payment_amount,
    ROUND(AVG(payment_amount), 2) AS average_payment
FROM gold.fact_payments
GROUP BY payment_method
ORDER BY total_payment_amount DESC;
```

```python
%sql
SELECT
    c.customer_id,
    c.customer_name,
    COUNT(f.order_id) AS total_orders,
    ROUND(SUM(f.order_amount), 2) AS total_spent,
    ROUND(AVG(f.order_amount), 2) AS average_order_value
FROM gold.fact_orders f
JOIN gold.dim_customer c
    ON f.customer_id = c.customer_id
GROUP BY
    c.customer_id,
    c.customer_name
ORDER BY total_spent DESC;
```

```python
%sql
SELECT
    t.tailor_id,
    t.tailor_name,
    COUNT(f.order_id) AS total_orders,
    SUM(f.completed_flag) AS completed_orders,
    SUM(f.pending_flag) AS pending_orders,
    ROUND(SUM(f.order_amount), 2) AS total_order_value,
    ROUND(AVG(f.order_amount), 2) AS average_order_value
FROM gold.fact_orders f
JOIN gold.dim_tailor t
    ON f.tailor_id = t.tailor_id
GROUP BY
    t.tailor_id,
    t.tailor_name
ORDER BY total_order_value DESC;
```

```python
%sql
SELECT
    order_status,
    COUNT(*) AS order_count,
    ROUND(SUM(order_amount), 2) AS total_value,
    ROUND(
        COUNT(*) * 100.0 /
        SUM(COUNT(*)) OVER (),
        2
    ) AS percentage_of_orders
FROM gold.fact_orders
GROUP BY order_status
ORDER BY order_count DESC;
```

```python
%sql
    
WITH customer_revenue AS (
    SELECT
        c.customer_id,
        c.customer_name,
        SUM(f.order_amount) AS total_spent
    FROM gold.fact_orders f
    JOIN gold.dim_customer c
        ON f.customer_id = c.customer_id
    GROUP BY
        c.customer_id,
        c.customer_name
)

SELECT
    customer_id,
    customer_name,
    ROUND(total_spent, 2) AS total_spent
FROM customer_revenue
ORDER BY total_spent DESC
LIMIT 5;
```

```python
%sql
    
WITH customer_revenue AS (
    SELECT
        c.customer_id,
        c.customer_name,
        SUM(f.order_amount) AS total_spent
    FROM gold.fact_orders f
    JOIN gold.dim_customer c
        ON f.customer_id = c.customer_id
    GROUP BY
        c.customer_id,
        c.customer_name
)

SELECT
    customer_id,
    customer_name,
    ROUND(total_spent, 2) AS total_spent,

    RANK() OVER (
        ORDER BY total_spent DESC
    ) AS spending_rank

FROM customer_revenue
ORDER BY spending_rank;
```

```python
%sql
SELECT
    t.tailor_id,
    t.tailor_name,
    COUNT(f.order_id) AS total_orders,
    SUM(f.completed_flag) AS completed_orders,
    SUM(f.pending_flag) AS pending_orders,
    ROUND(SUM(f.order_amount), 2) AS total_order_value,
    ROUND(AVG(f.order_amount), 2) AS average_order_value
FROM gold.fact_orders f
JOIN gold.dim_tailor t
    ON f.tailor_id = t.tailor_id
GROUP BY
    t.tailor_id,
    t.tailor_name
ORDER BY total_order_value DESC;
```

```python
%sql
SELECT
    order_status,
    COUNT(*) AS order_count,
    ROUND(SUM(order_amount), 2) AS total_value,
    ROUND(
        COUNT(*) * 100.0 /
        SUM(COUNT(*)) OVER (),
        2
    ) AS percentage_of_orders
FROM gold.fact_orders
GROUP BY order_status
ORDER BY order_count DESC;
```

```python
%sql
SELECT
    g.garment_type,
    COUNT(f.order_id) AS total_orders,
    SUM(f.completed_flag) AS completed_orders,
    SUM(f.pending_flag) AS pending_orders,
    ROUND(SUM(f.order_amount), 2) AS revenue
FROM gold.fact_orders f
JOIN gold.dim_garment g
    ON f.garment_id = g.garment_id
GROUP BY g.garment_type
ORDER BY revenue DESC;
```

```python
%sql
SELECT
    f.order_id,
    f.order_amount,
    p.payment_amount,
    ROUND(
        f.order_amount - COALESCE(p.payment_amount, 0),
        2
    ) AS outstanding_amount
FROM gold.fact_orders f
LEFT JOIN gold.fact_payments p
    ON f.order_id = p.order_id
ORDER BY outstanding_amount DESC;
```

```python
%sql
SELECT
    f.order_id,
    f.order_amount,
    COALESCE(p.payment_amount, 0) AS payment_amount,

    CASE
        WHEN COALESCE(p.payment_amount, 0) = f.order_amount
            THEN 'PAID'

        WHEN COALESCE(p.payment_amount, 0) < f.order_amount
            THEN 'PARTIALLY PAID'

        WHEN COALESCE(p.payment_amount, 0) > f.order_amount
            THEN 'OVERPAID'

        ELSE 'NO PAYMENT'
    END AS payment_status

FROM gold.fact_orders f
LEFT JOIN gold.fact_payments p
    ON f.order_id = p.order_id;
```

```python
%sql
    
WITH ranked_status AS (
    SELECT
        order_id,
        status,
        changed_on,
        tailor_id,

        ROW_NUMBER() OVER (
            PARTITION BY order_id
            ORDER BY changed_on DESC
        ) AS rn

    FROM silver.order_status_history
)

SELECT
    order_id,
    status AS latest_status,
    changed_on AS latest_status_time,
    tailor_id
FROM ranked_status
WHERE rn = 1
ORDER BY order_id;
```

```python
%sql
SELECT
    order_id,
    COUNT(*) AS number_of_status_changes
FROM silver.order_status_history
GROUP BY order_id
ORDER BY number_of_status_changes DESC;
```

```python
%sql
    
WITH status_times AS (
    SELECT
        order_id,

        MIN(changed_on) AS order_started_at,

        MAX(
            CASE
                WHEN status = 'COMPLETED'
                THEN changed_on
            END
        ) AS completed_at

    FROM silver.order_status_history

    GROUP BY order_id
)

SELECT
    order_id,
    order_started_at,
    completed_at,

    ROUND(
        (UNIX_TIMESTAMP(completed_at) -
         UNIX_TIMESTAMP(order_started_at)) / 86400.0,
        2
    ) AS turnaround_days

FROM status_times
WHERE completed_at IS NOT NULL
ORDER BY turnaround_days DESC;
```

```python
%sql
    
WITH status_times AS (
    SELECT
        order_id,
        MIN(changed_on) AS order_started_at,
        MAX(
            CASE
                WHEN status = 'COMPLETED'
                THEN changed_on
            END
        ) AS completed_at
    FROM silver.order_status_history
    GROUP BY order_id
),

turnaround AS (
    SELECT
        order_id,
        (
            UNIX_TIMESTAMP(completed_at) -
            UNIX_TIMESTAMP(order_started_at)
        ) / 86400.0 AS turnaround_days
    FROM status_times
    WHERE completed_at IS NOT NULL
)

SELECT
    ROUND(AVG(turnaround_days), 2) AS average_turnaround_days,
    ROUND(MIN(turnaround_days), 2) AS fastest_order_days,
    ROUND(MAX(turnaround_days), 2) AS slowest_order_days
FROM turnaround;
```

```python
%sql
    
WITH completion_dates AS (
    SELECT
        order_id,

        MAX(
            CASE
                WHEN status = 'COMPLETED'
                THEN changed_on
            END
        ) AS completed_at

    FROM silver.order_status_history
    GROUP BY order_id
)

SELECT
    f.order_id,
    c.customer_name,
    f.deadline_date,
    DATE(cd.completed_at) AS completed_date,

    CASE
        WHEN DATE(cd.completed_at) > f.deadline_date
            THEN 'OVERDUE'

        ELSE 'ON TIME'
    END AS delivery_status

FROM gold.fact_orders f

JOIN gold.dim_customer c
    ON f.customer_id = c.customer_id

JOIN completion_dates cd
    ON f.order_id = cd.order_id

WHERE cd.completed_at IS NOT NULL;
```
